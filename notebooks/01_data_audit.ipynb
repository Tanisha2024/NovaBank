```python
missing=(
    loan.isna()
    .sum()
    .sort_values(ascending=False)
    .to_frame('missing_count')
)
missing['missing_pct']=(
    missing['missing_count']/len(loan)*100
)
missing.head(20)
```

```python
# Search for columns relevant to our business questions

keywords = [
    "loan",
    "inc",
    "income",
    "fico",
    "dti",
    "delinq",
    "revol",
    "recover",
    "pymnt",
    "payment",
    "collection",
    "installment",
    "purpose",
    "emp",
    "grade",
    "term",
    "status"
]

relevant_columns = [
    col for col in loan.columns
    if any(keyword in col.lower() for keyword in keywords)
]

for col in relevant_columns:
    print(col)
```
